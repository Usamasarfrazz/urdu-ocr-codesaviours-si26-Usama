# Urdu OCR Project

## Week 1 Research

### What is OCR?
OCR (Optical Character Recognition) is a technology that converts text from images into editable digital text. It is widely used to digitize books, scanned documents, receipts, and printed materials.

### Why is Urdu OCR difficult?
Urdu OCR is more challenging than English OCR because Urdu letters are connected, their shapes change depending on their position in a word, and many letters look similar except for small dots. Urdu also has fewer publicly available datasets for training machine learning models.

### Real-World Applications
- Digitizing Urdu books, newspapers, and historical documents.
- Extracting text from Urdu forms, invoices, and government records automatically.

````python
!pip install Pillow arabic-reshaper python-bidi

from PIL import Image, ImageDraw, ImageFont
import arabic_reshaper
from bidi.algorithm import get_display
import os

os.makedirs("data/raw/synthetic", exist_ok=True)

texts = [
"پاکستان زندہ باد",
"تعلیم ہر انسان کا حق ہے",
"آج موسم خوشگوار ہے",
"کراچی پاکستان کا بڑا شہر ہے",
"لاہور ایک تاریخی شہر ہے",
"اسلام آباد خوبصورت دارالحکومت ہے",
"ہم سب پاکستانی ہیں",
"اردو ہماری قومی زبان ہے",
"سائنس ترقی کی بنیاد ہے",
"محنت کامیابی کی کنجی ہے",
"علم روشنی ہے",
"کتاب بہترین دوست ہے",
"پاکستان ایک خوبصورت ملک ہے",
"طلبہ محنت سے کامیاب ہوتے ہیں",
"وقت بہت قیمتی ہے",
"ایمان داری بہترین پالیسی ہے",
"پانی زندگی ہے",
"درخت ماحول کو بہتر بناتے ہیں",
"ہمیں صفائی رکھنی چاہیے",
"ہر دن نئی امید لاتا ہے"
]

for i, text in enumerate(texts):

    reshaped = arabic_reshaper.reshape(text)
    bidi_text = get_display(reshaped)

    img = Image.new("RGB",(700,120),"white")
    draw = ImageDraw.Draw(img)

    draw.text((20,40),bidi_text,fill="black")

    img.save(f"data/raw/synthetic/img_{i+1:03}.png")

print("20 Synthetic Images Generated")
````

````python
import csv
import os

# Ensure the data directory exists for labels.csv
os.makedirs('data', exist_ok=True)

# The 'texts' list is available from the previous cell (7MWHwe9HdfmP)
# Re-creating texts list here to ensure the code is self-contained if run independently
texts = [
    "پاکستان زندہ باد",
    "تعلیم ہر انسان کا حق ہے",
    "آج موسم خوشگوار ہے",
    "کراچی پاکستان کا بڑا شہر ہے",
    "لاہور ایک تاریخی شہر ہے",
    "اسلام آباد خوبصورت دارالحکومت ہے",
    "ہم سب پاکستانی ہیں",
    "اردو ہماری قومی زبان ہے",
    "سائنس ترقی کی بنیاد ہے",
    "محنت کامیابی کی کنجی ہے",
    "علم روشنی ہے",
    "کتاب بہترین دوست ہے",
    "پاکستان ایک خوبصورت ملک ہے",
    "طلبہ محنت سے کامیاب ہوتے ہیں",
    "وقت بہت قیمتی ہے",
    "ایمان داری بہترین پالیسی ہے",
    "پانی زندگی ہے",
    "درخت ماحول کو بہتر بناتے ہیں",
    "ہمیں صفائی رکھنی چاہیے",
    "ہر دن نئی امید لاتا ہے"
]

# Create a dictionary to map synthetic image filenames to their texts
synthetic_labels_map = {}
for i, text_content in enumerate(texts):
    image_filename_key = f'data/raw/synthetic/img_{i+1:03}.png'
    synthetic_labels_map[image_filename_key] = text_content

all_image_labels = []
base_dir = 'data/raw'

# Walk through all subdirectories in data/raw to find image files
for root, _, files in os.walk(base_dir):
    for file in files:
        if file.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp')):
            full_path = os.path.join(root, file).replace(os.sep, '/') # Use forward slashes for consistency

            # Check if it's a synthetic image and get its text
            if full_path in synthetic_labels_map:
                text_label = synthetic_labels_map[full_path]
            else:
                # For other images, initialize text as empty
                text_label = ''

            all_image_labels.append({'image': full_path, 'text': text_label})

# Sort the labels by image path for consistency
all_image_labels.sort(key=lambda x: x['image'])

with open('data/labels.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['image', 'text'])
    writer.writeheader()
    writer.writerows(all_image_labels)

print(f'Labels file created with {len(all_image_labels)} entries, covering all images found in {base_dir}')
````
